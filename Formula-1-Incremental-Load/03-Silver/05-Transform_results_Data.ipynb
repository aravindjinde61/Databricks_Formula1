```python
bronze_table = f"{catalog_name}.{bronze_schema}.results"
silver_table = f"{catalog_name}.{silver_schema}.results"
```

```python
from pyspark.sql import functions as f
```

```python
# To use reader api options 
results_df = spark.read.table(bronze_table).filter((f.col("batch_id") == v_batch_id))

# results_df = spark.table(bronze_table) - without any reader options
# Both returns the same results
```

```python
results_dropped_df = results_df.drop(f.col("url"))
```

```python
# For column rename we can use ALIAS funcrion or also the below function

column_rename = {
    "date" : "race_date",
    "grid" : "grid_position",
    "laps" : "completed_laps",
    "number" : "car_number",
    "position" : "finish_position",
    "constructorId" : "constructor_id",
    "driverId" : "driver_id",
    "raceName" : "race_name",
    "positionText" : "finish_position_text"
    }

results_renamed_df = results_dropped_df.withColumnsRenamed(column_rename)
```

```python
# To Drop duplicates and Null values in a column - distinct(), dropDuplicates() & filter()

results_filtered_df = results_renamed_df.filter(
    (
        (f.col("driver_id").isNotNull()) & (f.col("constructor_id").isNotNull()) & (f.col("round").isNotNull()) & (f.col("season").isNotNull())
    )
)

results_valid_df = results_filtered_df.dropDuplicates(["driver_id","constructor_id","round","season"])
```

```python
results_final_df = results_valid_df.withColumn("race_name", f.initcap("race_name"))
```

```python
columns = [
    "race_name",
    "race_date",
    "grid_position",
    "completed_laps",
    "car_number",
    "points",
    "finish_position",
    "finish_position_text",
    "status",
    "ingestion_timestamp",
    "source_file",
    "batch_id"
]

write_to_silver(results_final_df, silver_table, "trg.season = src.season AND trg.round = src.round AND trg.constructor_id = src.constructor_id AND trg.driver_id = src.driver_id", columns)
```
